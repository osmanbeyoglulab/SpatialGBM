```python
import drug2cell as d2c
import scanpy as sc
import os, warnings
warnings.simplefilter("ignore")
from glob import glob
import stan
```

```python
path = 'results'
if not os.path.exists(path):
    os.mkdir(path)
```

## Ravi et al.

```python
sample_list = []
for x in glob('data/Ravi_processed/gbm_qc/*'):
    sample_list.append(x.split(r'/')[-1][:-5])
sample_list.sort()
sample_list
```

```python
adatas = dict()
for sample in sample_list:
    adatas[sample] = sc.read_h5ad(f"data/Ravi_processed/gbm_qc/{sample}.h5ad")
    sc.pp.normalize_total(adatas[sample])
    sc.pp.log1p(adatas[sample])
    d2c.score(adatas[sample], use_raw=True)
```

```python
adatas_d2c = dict()
for sample in sample_list:
    adata = adatas[sample].copy()
    adata.uns['drug2cell'].uns['spatial'] = adata.uns['spatial']
    adatas_d2c[sample] = adata.uns['drug2cell']
    del adata
```

```python
path = 'results/ravi_d2c'
if not os.path.exists(path):
    os.mkdir(path)
    
for sample in sample_list:
    adatas_d2c[sample].write(f"{path}/{sample}.h5ad")
```
