```python
import matplotlib.pyplot as plt
import pandas as pd
import scanpy as sc
import squidpy as sq
import os, warnings 
warnings.filterwarnings('ignore') 

from setup import *
from glob import glob
import stan
```

```python
def plot_spatial_activity(adata, sample, tf, path='plots_website/transcription'):
    title = f"{sample}\n{tf} activity\n Moran's I = {moran_df.loc[tf, sample]:.2f}"
    fig, ax = plt.subplots(1, 1, figsize=(2, 2.1), dpi=dpi)
    sc.pl.spatial(adata, color=tf, alpha_img=0, ax=ax, show=False, cmap="plasma", 
                  legend_fontsize=fontsize, size=1.8)
    cbar = ax.collections[0].colorbar
    cbar.ax.tick_params(labelsize=fontsize)
    ax.set_title(title, fontsize=fontsize)
    ax.set_xlabel("")
    ax.set_ylabel("")
    plt.tight_layout(pad=1)
    
    # Save the plot
    plt.savefig(f'{path}/{sample}.png')
    plt.close()
```

```python
def plot_violin_activity(adata, sample, tf, path='plots_website/transcription_violin'):
    fig, ax = plt.subplots(1, 1, figsize=(3, 2.1), dpi=dpi)
    sc.pl.violin(adata, keys=tf, groupby='annotation', show=False, 
                 palette=metaprogram_palette, ax=ax, rotation=90)
    ax.set_title(sample, fontsize=fontsize)
    ax.set_ylabel('{} activity'.format(tf), fontsize=fontsize)
    plt.xticks(fontsize=fontsize)
    plt.yticks(fontsize=fontsize)
    plt.tight_layout(pad=1)
    
    # Save the plot
    plt.savefig(f'{path}/{sample}.png')
    plt.close()
```

## Ravi et al.

```python
sample_list = []
for x in glob('data/Ravi_processed/gbm_qc/*'):
    sample_list.append(x.split(r'/')[-1][:-5])
sample_list.sort()
```

```python
adatas = dict()
for sample in sample_list:
    adata = sc.read_h5ad("results/ravi_tfa/{}.h5ad".format(sample))
    adatas[sample] = stan.get_activity(adata, key='tfa_stan')
```

```python
fname = 'dataframes/morani_ravi/tfa_per_sample.csv'
if os.path.isfile(fname):
    moran_df = pd.read_csv(fname, index_col=0)
else:
    moran_df = stan.compute_morans_i(adatas)
    moran_df.to_csv(fname)
```

```python
# Spatial plots
path = 'plots_website/ravi_transcription'
if not os.path.exists(path):
    os.mkdir(path)

for sample in sample_list[:3]: ## remove [:3]
    adata = adatas[sample]
    for tf in adata.var.index[:3]: ## remove [:3]
        child_path = f'{path}/{tf}'
        if not os.path.exists(child_path):
            os.mkdir(child_path)
        plot_spatial_activity(adata, sample, tf, path=child_path)
```

```python
# Violin plots
path = 'plots_website/ravi_transcription_violin'
if not os.path.exists(path):
    os.mkdir(path)
    
for sample in sample_list[:3]: ## remove [:3]
    adata = adatas[sample]
    for tf in adata.var.index[:3]: ## remove [:3]
        child_path = f'{path}/{tf}'
        if not os.path.exists(child_path):
            os.mkdir(child_path)
        plot_violin_activity(adata, sample, tf, path=child_path)
```
