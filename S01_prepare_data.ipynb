```python
import numpy as np
import pandas as pd
import scanpy as sc

import anndata, os, warnings
warnings.simplefilter("ignore")
from glob import glob
```

## Ravi et al.

Download the dataset: https://drive.google.com/file/d/1YFOwEDFLpyxG6WX243FNA7lnrVSaiVL-/view?usp=sharing

```python
sample_list = []
for x in glob('Spatial_Glioma/Inputs/general/GBM_data/*'):
    sample_list.append(x.split('/')[-1])
sample_list.sort()
sample_list
```

```python
path = 'data/Ravi_processed/gbm_st/'
if not os.path.exists(path):
    os.mkdir(path)
```

### Add metadata 
Run the R code first.

```python
for sample in sample_list:
    path_in = f'Spatial_Glioma/Inputs/general/GBM_data/{sample}/outs'
    adata = sc.read_visium(path_in)
    adata.var_names_make_unique()
    
    path_meta = f'data/Ravi_processed/gbm_meta/{sample}.csv'
    anno = pd.read_csv(path_meta, index_col=1)
    
    obs_names = np.intersect1d(anno.index, adata.obs_names)
    adata = adata[obs_names]
    adata.obs['annotation'] = anno['spot_type_meta_new']
    print(f'{sample}: {len(obs_names)}')
    
    adata.write(f'{path}/{sample}.h5ad')
```

### Filter dataset

```python
for sample in sample_list:
    sc.pp.filter_genes(adatas[sample], min_cells=5)
    sc.pp.filter_cells(adatas[sample], min_counts=1000)
    adatas[sample].layers['raw'] = adatas[sample].X
    adatas[sample].obs['ncounts'] = adatas[sample].to_df('raw').T.sum()
```

```python
path = 'data/Ravi_processed/gbm_qc/'
if not os.path.exists(path):
    os.mkdir(path)
    
for sample in sample_list:
    adatas[sample].write(f'{path}/{sample}.h5ad')
```

## Sonpatki et al.

```python
sample_list = []
for x in glob('data/Sonpatki_raw/*'):
    sample_list.append(x.split('/')[-1].split('_')[0])
sample_list.sort()
sample_list
```
