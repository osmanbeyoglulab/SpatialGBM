```python
import matplotlib.pyplot as plt
import pandas as pd
import scanpy as sc
import squidpy as sq
import os, warnings 
warnings.filterwarnings('ignore') 

from setup import *
from glob import glob
import stan
```

```python
# Gene expression (lop1p normalized) in spatial coordinates
def plot_spatial_expression(adata, sample, gene, path='plots_website/mrna'):
    title = f"{sample}\n{gene} expression\n Moran's I = {moran_df.loc[gene, sample]:.2f}"
    fig, ax = plt.subplots(1, 1, figsize=(2, 2.1), dpi=dpi)
    sc.pl.spatial(adata, color=gene, alpha_img=0, ax=ax, show=False, cmap="viridis", 
                  legend_fontsize=fontsize, size=1.8,
                  vmin=adata.X.min(), vmax=adata.X.max())
    cbar = ax.collections[0].colorbar
    cbar.ax.tick_params(labelsize=fontsize)
    ax.set_title(title, fontsize=fontsize)
    ax.set_xlabel("")
    ax.set_ylabel("")
    plt.tight_layout(pad=1)
    
    # Save the plot
    plt.savefig(f'{path}/{sample}.pdf')
    plt.close()
```

```python
# Gene expression (lop1p normalized) by metaprogram
def plot_violin_expression(adata, sample, gene, path='plots_website/mrna_violin'):
    fig, ax = plt.subplots(1, 1, figsize=(3, 2.1), dpi=dpi)
    sc.pl.violin(adata, keys=gene, groupby='annotation', show=False, 
                 palette=metaprogram_palette, ax=ax, rotation=90)
    ax.set_title(sample, fontsize=fontsize)
    ax.set_ylabel(f'{gene} expression', fontsize=fontsize)
    plt.xticks(fontsize=fontsize)
    plt.yticks(fontsize=fontsize)
    plt.tight_layout(pad=1)
    
    # Save the plot
    plt.savefig(f'{path}/{sample}.png')
    plt.close()
```

## Ravi et al.

```python
sample_list = []
for x in glob('data/Ravi_processed/gbm_qc/*'):
    sample_list.append(x.split(r'/')[-1][:-5])
sample_list.sort()
```

```python
adatas = dict()
genes = dict()
for sample in sample_list:
    adatas[sample] = sc.read_h5ad(f"data/Ravi_processed/gbm_qc/{sample}.h5ad")
    genes[sample] = filter_genes(adatas[sample])
    sc.pp.normalize_total(adatas[sample])
    sc.pp.log1p(adatas[sample])
```

```python
fname = 'dataframes/morani_ravi/mrna_per_sample.csv'
if os.path.isfile(fname):
    moran_df = pd.read_csv(fname, index_col=0)
else:
    moran_df = stan.compute_morans_i(adatas)
    moran_df.to_csv(fname)
```

```python
# Spatial plots
path = 'plots_website/ravi_mrna'
if not os.path.exists(path):
    os.mkdir(path)

for sample in sample_list[:3]: ## remove [:3]
    adata = adatas[sample]
    for gene in genes[sample][:3]: ## remove [:3]
        child_path = f'{path}/{gene}'
        if not os.path.exists(child_path):
            os.mkdir(child_path)
        plot_spatial_expression(adata, sample, gene, path=child_path)
```

```python
# Violin plots
path = 'plots_website/ravi_mrna_violin'
if not os.path.exists(path):
    os.mkdir(path)

for sample in sample_list[:3]: ## remove [:3]
    adata = adatas[sample]
    for gene in genes[sample][:3]: ## remove [:3]
        child_path = f'{path}/{gene}'
        if not os.path.exists(child_path):
            os.mkdir(child_path)
        plot_violin_expression(adata, sample, gene, path=child_path)
```

```python
all_genes = set()
for gene_list in genes.values():
    all_genes.update(gene_list)
all_genes = sorted(all_genes)

with open('dataframes/ravi_selected_genes.txt', 'w') as f:
    for gene in all_genes:
        f.write(f"{gene}\n")
```
