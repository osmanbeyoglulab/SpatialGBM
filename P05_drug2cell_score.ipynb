```python
import matplotlib.pyplot as plt
import pandas as pd
import scanpy as sc
import squidpy as sq
import os, warnings 
warnings.filterwarnings('ignore') 

from setup import *
from glob import glob
import stan
```

```python
def plot_spatial_score(adata, sample, drug, path='plots_website/drug'):
    d1, d2 = drug.split('|')
    d3 = split_at_natural_break(d2)
    title = f"{sample}\n{d1} score\n{d3}\nMoran's I = {moran_df.loc[drug, sample]:.2f}"
    
    fig, ax = plt.subplots(1, 1, figsize=(2.4, 2.4), dpi=dpi)
    sc.pl.spatial(adata, color=drug, alpha_img=0, ax=ax, show=False, cmap="inferno", 
                  legend_fontsize=fontsize, size=1.8)
    cbar = ax.collections[0].colorbar
    cbar.ax.tick_params(labelsize=fontsize)
    ax.set_title(title, fontsize=fontsize)
    ax.set_xlabel("")
    ax.set_ylabel("")
    plt.tight_layout(pad=1)
    
    # Save the plot
    plt.savefig(f'{path}/{sample}.png')
    plt.close()
```

```python
def plot_violin_score(adata, sample, drug, path='plots_website/drug_violin'):
    fig, ax = plt.subplots(1, 1, figsize=(3, 3), dpi=dpi)
    sc.pl.violin(adata, keys=drug, groupby='annotation', show=False, 
                 palette=metaprogram_palette, ax=ax, rotation=90)
    ax.set_title(sample, fontsize=fontsize)
    d1, d2 = drug.split('|')
    d3 = split_at_natural_break(d2)
    ax.set_ylabel('{} score\n{}'.format(d1, d3), fontsize=fontsize)
    plt.xticks(fontsize=fontsize)
    plt.yticks(fontsize=fontsize)
    plt.tight_layout(pad=1)
    
    # Save the plot
    plt.savefig(f'{path}/{sample}.png')
    plt.close()
```

## Ravi et al.

```python
sample_list = []
for x in glob('data/Ravi_processed/gbm_qc/*'):
    sample_list.append(x.split(r'/')[-1][:-5])
sample_list.sort()
```

```python
adatas = dict()
for sample in sample_list:
    adatas[sample] = sc.read_h5ad(f"results/ravi_d2c/{sample}.h5ad")
```

```python
fname = 'dataframes/morani_ravi/d2c_per_sample.csv'
if os.path.isfile(fname):
    moran_df = pd.read_csv(fname, index_col=0)
else:
    moran_df = stan.compute_morans_i(adatas)
    moran_df.to_csv(fname)
```

```python
# Spatial plots
path = 'plots_website/ravi_drug'
if not os.path.exists(path):
    os.mkdir(path)

for sample in sample_list[:3]: ## remove [:3]
    adata = adatas[sample]
    for drug in adata.var.index[:3]: ## remove [:3]
        child_path = f'{path}/{drug}'
        if not os.path.exists(child_path):
            os.mkdir(child_path)
        plot_spatial_score(adata, sample, drug, path=child_path)
```

```python
# Violin plots
path = 'plots_website/ravi_drug_violin'
if not os.path.exists(path):
    os.mkdir(path)
    
for sample in sample_list[:3]: ## remove [:3]
    adata = adatas[sample]
    for drug in adata.var.index[:3]: ## remove [:3]
        child_path = f'{path}/{drug}'
        if not os.path.exists(child_path):
            os.mkdir(child_path)
        plot_violin_score(adata, sample, drug, path=child_path)
```
