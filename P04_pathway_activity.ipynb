```python
import matplotlib.pyplot as plt
import pandas as pd
import scanpy as sc
import squidpy as sq
import os, warnings 
warnings.filterwarnings('ignore') 

from setup import *
from glob import glob
import stan
```

```python
def plot_spatial_activity(adata, sample, pathway, path='plots_website/pathway'):
    title = f"{sample}\n{hallmarks[pathway]} activity\n Moran's I = {moran_df.loc[pathway, sample]:.2f}"
    fig, ax = plt.subplots(1, 1, figsize=(2, 2.1), dpi=dpi)
    sc.pl.spatial(adata, color=pathway, alpha_img=0, ax=ax, show=False, cmap="rocket", 
                  legend_fontsize=fontsize, size=1.8)
    cbar = ax.collections[0].colorbar
    cbar.ax.tick_params(labelsize=fontsize)
    ax.set_title(title, fontsize=fontsize)
    ax.set_xlabel("")
    ax.set_ylabel("")
    plt.tight_layout(pad=1)
    
    # Save the plot
    plt.savefig(f'{path}/{sample}.png')
    plt.close()
```

```python
def plot_violin_activity(adata, sample, pathway, path='plots_website/pathway_violin'):
    fig, ax = plt.subplots(1, 1, figsize=(3, 2.1), dpi=dpi)
    sc.pl.violin(adata, keys=pathway, groupby='annotation', show=False, 
                 palette=metaprogram_palette, ax=ax, rotation=90)
    ax.set_title(sample, fontsize=fontsize)
    ax.set_ylabel('{}\nactivity'.format(hallmarks[pathway]), fontsize=fontsize)
    plt.xticks(fontsize=fontsize)
    plt.yticks(fontsize=fontsize)
    plt.tight_layout(pad=1)
    
    # Save the plot
    plt.savefig(f'{path}/{sample}.png')
    plt.close()
```

## Ravi et al.

```python
sample_list = []
for x in glob('data/Ravi_processed/gbm_qc/*'):
    sample_list.append(x.split(r'/')[-1][:-5])
sample_list.sort()
```

```python
adatas = dict()
for sample in sample_list:
    adata = sc.read_h5ad(f"results/ravi_pwa/{sample}.h5ad")
    adatas[sample] = stan.get_activity(adata, key='pwa_stan')
```

```python
fname = 'dataframes/morani_ravi/pwa_per_sample.csv'
if os.path.isfile(fname):
    moran_df = pd.read_csv(fname, index_col=0)
else:
    moran_df = stan.compute_morans_i(adatas)
    moran_df.to_csv(fname)
```

```python
# Spatial plots
path = 'plots_website/ravi_pathway'
if not os.path.exists(path):
    os.mkdir(path)

for sample in sample_list[:3]: ## remove [:3]
    adata = adatas[sample]
    for pathway in adata.var.index[:3]: ## remove [:3]
        child_path = f'{path}/{pathway}'
        if not os.path.exists(child_path):
            os.mkdir(child_path)
        plot_spatial_activity(adata, sample, pathway, path=child_path)
```

```python
# Violin plots
path = 'plots_website/ravi_pathway_violin'
if not os.path.exists(path):
    os.mkdir(path)
    
for sample in sample_list[:3]: ## remove [:3]
    adata = adatas[sample]
    for pathway in adata.var.index[:3]: ## remove [:3]
        child_path = f'{path}/{pathway}'
        if not os.path.exists(child_path):
            os.mkdir(child_path)
        plot_violin_activity(adata, sample, pathway, path=child_path)
```
